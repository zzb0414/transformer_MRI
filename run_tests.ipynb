## Performance Benchmark

Measure the forward pass time for the full model.

```python
import time

def benchmark_model(device='cpu', num_runs=10):
    """Benchmark transformer forward pass."""
    
    W, H = 64, 64  # Standard size
    d_model = 768
    N = 4
    num_heads = 8
    seq_len = W * H
    omega = torch.ones(1, seq_len)
    
    # Correct: np.mgrid[0:H, 0:W] produces shape [2, H, W]
    pos = np.mgrid[0:H, 0:W]
    model = transformer(
        input_channel=2,
        output_channel=d_model,
        bias=False,
        pos=pos,
        d_model=d_model,
        omega=omega,
        N=N,
        num_heads=num_heads,
        W=W,
        H=H,
        activition=None,
        approach='sine',
        dropout_emb=0.0,
        dropout_enc=0.0,
        dropout_dec=0.0
    ).to(device)
    
    model.eval()
    
    # Warmup
    ksp = torch.randn(1, seq_len, 2).to(device)
    omega = torch.ones(1, seq_len).to(device)
    with torch.no_grad():
        for _ in range(3):
            _ = model(ksp)
    
    # Benchmark
    times = []
    with torch.no_grad():
        for _ in range(num_runs):
            start = time.time()
            _ = model(ksp)
            if device == 'cuda':
                torch.cuda.synchronize()
            times.append(time.time() - start)
    
    avg_time = sum(times) / len(times)
    print(f"\nDevice: {device.upper()}")
    print(f"Image size: {W}x{H} ({seq_len} points)")
    print(f"Model: d_model={d_model}, N={N}, heads={num_heads}")
    print(f"Average forward pass time: {avg_time*1000:.2f} ms")
    print(f"Throughput: {1/avg_time:.2f} images/sec")
    
    return avg_time

# CPU benchmark
cpu_time = benchmark_model('cpu', num_runs=5)

# GPU benchmark (if available)
if torch.cuda.is_available():
    gpu_time = benchmark_model('cuda', num_runs=10)
    print(f"\nGPU Speedup: {cpu_time/gpu_time:.2f}x")
else:
    print("\nCUDA not available for GPU benchmark.")
```
